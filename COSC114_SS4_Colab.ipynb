**`and` and `or` return an actual operand, without converting the result to `True` or `False`.**

| Operator | Decision rule |
| --- | --- |
| `or` | Return the first truthy operand. If none is truthy, return the last operand. |
| `and` | Return the first falsy operand. If none is falsy, return the last operand. |

Evaluation runs from left to right and stops as soon as the result is decided.
This is **short-circuiting**. The returned operand keeps its value and type.


````python
a = 2
b = 3
print(a and b)
print(a or b)

a = True
b = 1
print(a or b)
print(a and b)

print(0 and 9)
print('' or 'guest')
````

````python
a = 100
b = a  # Explicit alias: a and b refer to the same object.
print(a == b, a is b)
print(id(a), id(b))  # Actual numbers vary between runs.
a = 'KU'
print(a, type(a))
print(b, type(b))  # Reassignment of a did not change b.
print(a is b)
number = 100
decimal_number = 100.0
print(number == decimal_number, number is decimal_number)
````

#### Slide 7B — Stack memory

<div class="cosc-practice-slide">
<p>The <strong>call stack</strong> keeps track of running function calls.</p>
<ul>
<li>Each call gets a frame on the stack that holds its local names, which are references to objects.</li>
<li>When the function finishes, its frame is removed automatically.</li>
<li>Frames are added and removed in order: the last call to start is the first to finish.</li>
<li>Python manages this for you. It is fast but temporary: a frame lasts only while its call is running.</li>
</ul>
<p><strong>In simple terms:</strong> the stack is short-lived bookkeeping for function calls.</p>
</div>


````python
def func():
    # While func() runs, the names a, b and c belong to its frame on the call stack.
    a = 20
    b = []
    c = ""
    print(a, b, repr(c))

func()  # When func() returns, its frame and these local names are gone.
````

#### Slide 7C — Heap memory

<div class="cosc-practice-slide">
<p>The <strong>heap</strong> is where the objects themselves live: numbers, strings, lists, dictionaries.</p>
<ul>
<li>Python creates objects on the heap at runtime, in no fixed order.</li>
<li>Names hold only a reference to the object, never the object itself.</li>
<li>An object can be shared by several names and can outlive the function that created it.</li>
<li>When no reference is left, Python frees the object (reference counting and garbage collection).</li>
</ul>
<p><strong>In simple terms:</strong> the heap stores the values; names on the stack store directions to them.</p>
</div>


````python
# The list object is created on the heap; the name a refers to it.
a = [0] * 10
print(a)
````

**Conceptual model.** CPython keeps every object on its heap and stores a function's local names in its frame.
Exact storage is an implementation detail. Adapted from GeeksforGeeks, *Memory Management in Python*.

#### Solution, step by step — `x = 8`, `y = 25`, `z = 3`

`5 < x * 2 <= y - 5 and not z + x // 2 > 10 or y % 4 == 0`

**Parenthesised:** `((5 < x * 2 <= y - 5) and (not (z + x // 2 > 10))) or (y % 4 == 0)`

**1. Arithmetic first** (`*`, `//`, `%`, then `+`, `-`)

| Part | Substitute | Value |
|---|---|---|
| `x * 2` | `8 * 2` | `16` |
| `y - 5` | `25 - 5` | `20` |
| `x // 2` | `8 // 2` | `4` |
| `z + x // 2` | `3 + 4` | `7` |
| `y % 4` | `25 % 4` | `1` |

Expression: `5 < 16 <= 20 and not 7 > 10 or 1 == 0`

**2. Comparisons** (`<`, `<=`, `>`, `==`)

| Part | Meaning | Value |
|---|---|---|
| `5 < 16 <= 20` | `(5 < 16) and (16 <= 20)` (a chained comparison) | `True` |
| `7 > 10` | | `False` |
| `1 == 0` | | `False` |

Expression: `True and not False or False`

**3. Logical `not`:** `not False` → `True`

Expression: `True and True or False`

**4. Logical `and`:** `True and True` → `True`

Expression: `True or False`

**5. Logical `or`:** `True or False` → `True`

**Output:** `True`

> **How Python actually runs it.** The stages above find the value by hand. Python uses precedence only to
> *group* the expression, then evaluates left to right. The left side of `or` is already `True`, so `or`
> short-circuits: `y % 4` and `1 == 0` are never computed. The answer is the same, `True`.
