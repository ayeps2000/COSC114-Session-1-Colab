```python
# @title Solution { display-mode: "form" }

# A) Swap — Python does it in one line, but the three-line version shows why.
a, b = 3, 7
a, b = b, a
print("A) a =", a, " b =", b)

# B) Even or odd — one decision, so one if/else.
number = 7
if number % 2 == 0:
    print("B)", number, "is even")
else:
    print("B)", number, "is odd")

# C) Sum of the first N — a running total, exactly like Example 1.
n = 5
total = 0
for i in range(1, n + 1):
    total = total + i
print("C) sum of the first", n, "naturals =", total)
```

<h2 style="color:#2457a6;font-size:22px;margin:0 0 0.6em;">Exercise 1: Scholarship eligibility</h2>
<ol><li><strong>Decompose:</strong> input is N marks; processing is one three-way decision per mark; output is three counts.</li><li><strong>Abstract:</strong> names and IDs do not matter &mdash; only the score and the two thresholds 60 and 85.</li><li><strong>Algorithm:</strong> the bands do not overlap, so this is one <code style="background:#eef3fa;color:#173d72;border-radius:4px;padding:0.05em 0.25em;">if</code> / <code style="background:#eef3fa;color:#173d72;border-radius:4px;padding:0.05em 0.25em;">elif</code> / <code style="background:#eef3fa;color:#173d72;border-radius:4px;padding:0.05em 0.25em;">else</code>, not three separate ifs.</li></ol>
<p><strong>Target:</strong> with <code style="background:#eef3fa;color:#173d72;border-radius:4px;padding:0.05em 0.25em;">[92, 58, 71, 85, 60, 44]</code> the counts are Fail <code style="background:#eef3fa;color:#173d72;border-radius:4px;padding:0.05em 0.25em;">2</code>, Pass <code style="background:#eef3fa;color:#173d72;border-radius:4px;padding:0.05em 0.25em;">2</code>, Eligible <code style="background:#eef3fa;color:#173d72;border-radius:4px;padding:0.05em 0.25em;">2</code>.</p>
</div>
</details>

```python
# @title Solution { display-mode: "form" }

marks = [92, 58, 71, 85, 60, 44]

fail = 0
passed = 0
eligible = 0

for score in marks:
    if score >= 85:              # check the highest band first
        eligible = eligible + 1
    elif score >= 60:            # reached only when score < 85
        passed = passed + 1
    else:
        fail = fail + 1

print("Fail:", fail, " Pass:", passed, " Eligible:", eligible)
```
