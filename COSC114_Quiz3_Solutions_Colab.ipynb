## Q.1 Slicing a list [10 marks]

Completed program. Run it and check each blank.

```python
temps = [31, 29, 35, 38, 40, 36, 33]
print(temps[1])       # the second reading
print(temps[:3])      # the first three readings
print(temps[-2:])     # the last two readings
print(temps[4:-7])    # What is the output of this line?
print(temps[4:-7:-3]) # What is the output of this line?
```

| Blank | Answer | Why | Class slide |
|:-:|---|---|---|
| 1 | `1` | Indexes start at 0, so the second reading is at index 1. | SS6 · 10 |

## Q.3 Dictionary values [10 marks]

Completed program. Run it and check each blank.

```python
prices = {"pen": 2.5, "book": 30.0, "bag": 75.0}
print(prices["book"])
total = 0
for item in prices:
    total = total + prices[item]
print(total)
print(75.0 in prices.values())
for item, cost in prices.items():
    print(item, "->", cost)
```

| Blank | Answer | Why | Class slide |

| 1 | `"book"` | A value is read with its key in square brackets. The key is a string. | SS6 · 19 |
| 2 | `prices` | A for loop over a dictionary gives its keys. | SS6 · 28 |

## Q.5 Square list entries [10 marks]

Model answer. Run it and compare with your program. Equivalent correct programs are acceptable.

```python
def squareEach(nums):
    for index in range(len(nums)):
        nums[index] = nums[index] ** 2

numbers = [2, -3, 1.5, 0]
squareEach(numbers)
print(numbers)
```

**Expected output**

<pre style="background:white;border:1px solid #d9e2ef;border-radius:6px;padding:8px 12px;font-family:Menlo,Consolas,monospace;font-size:16px;white-space:pre-wrap;">[4, 9, 2.25, 0]</pre>

| Marks | What the answer must do | Class slide |
|:-:|---|---|
| 2 | Function header squareEach(nums) with one list parameter. | SS7 · 37 |
| 3 | A loop over range(len(nums)) processes every entry by index. | SS7 · 37 |
| 3 | Each value is squared and assigned back to its own index. | SS7 · 5 |
| 2 | The list is changed in place: the caller sees the changes, no return needed. | SS7 · 37 |
