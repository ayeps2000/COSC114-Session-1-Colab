<details><summary style="cursor:pointer;color:#198754;font-weight:700;">Show the solution</summary>

#### Solution, step by step — `x = 8`, `y = 25`, `z = 3`

`5 < x * 2 <= y - 5 and not z + x // 2 > 10 or y % 4 == 0`

**Parenthesised:** `((5 < x * 2 <= y - 5) and (not (z + x // 2 > 10))) or (y % 4 == 0)`

**1. Arithmetic first** (`*`, `//`, `%`, then `+`, `-`)

| Part | Substitute | Value |
|---|---|---|
| `x * 2` | `8 * 2` | `16` |
| `y - 5` | `25 - 5` | `20` |
| `x // 2` | `8 // 2` | `4` |
| `z + x // 2` | `3 + 4` | `7` |
| `y % 4` | `25 % 4` | `1` |

Expression: `5 < 16 <= 20 and not 7 > 10 or 1 == 0`

**2. Comparisons** (`<`, `<=`, `>`, `==`)

| Part | Meaning | Value |
|---|---|---|
| `5 < 16 <= 20` | `(5 < 16) and (16 <= 20)` (a chained comparison) | `True` |
| `7 > 10` | | `False` |
| `1 == 0` | | `False` |

Expression: `True and not False or False`

**3. Logical `not`:** `not False` → `True`

Expression: `True and True or False`

**4. Logical `and`:** `True and True` → `True`

Expression: `True or False`

**5. Logical `or`:** `True or False` → `True`

**Output:** `True`

> **How Python actually runs it.** The stages above find the value by hand. Python uses precedence only to
> *group* the expression, then evaluates left to right. The left side of `or` is already `True`, so `or`
> short-circuits: `y % 4` and `1 == 0` are never computed. The answer is the same, `True`.

</details>

### Exercise 4 — Weekday

*SS4, slide 36*

<div style="border-left:6px solid #198754;background:#f2f8f5;border-radius:10px;padding:14px 18px;margin:6px 0;"><p>Map 1 to Monday through 7 to Sunday using if-elif-else. Print an error for invalid text or an out-of-range number.</p></div>


````python
text = input("Enter a day number (1-7): ").strip()
# TODO: validate decimal text, then choose the weekday.
pass
````

````python
# @title Solution { display-mode: "form" }
text = input("Enter a day number (1-7): ").strip()
if not text.isdecimal():
    print("Invalid day: enter 1-7")
else:
    day = int(text)
    if day == 1:
        print("Monday")
    elif day == 2:
        print("Tuesday")
    elif day == 3:
        print("Wednesday")
    elif day == 4:
        print("Thursday")
    elif day == 5:
        print("Friday")
    elif day == 6:
        print("Saturday")
    elif day == 7:
        print("Sunday")
    else:
        print("Invalid day: enter 1-7")
````

### Exercise 5 — Sort three integers

*SS4, slide 36*

<div style="border-left:6px solid #198754;background:#f2f8f5;border-radius:10px;padding:14px 18px;margin:6px 0;"><p>Assume valid integer input. Arrange three integers in ascending order using only conditionals: no loops, sorted(), sort(), min(), or max(). Include equal and negative values in your checks.</p></div>


````python
a = int(input("First integer: "))
b = int(input("Second integer: "))
c = int(input("Third integer: "))
# TODO: compare pairs and swap values with a temporary variable.
pass
````

````python
def greet():
    x = 1
    y = x + 1
    hi()
    print(x, y)

def hi():
    x = 3
    y = x + 1
    return x

greet()
````

````python
# @title Solution { display-mode: "form" }
def divisible(n, m):
    result = []
    for i in range(1, n + 1):
        if i % m == 0:
            result.append(i)
    return result

print(divisible(20, 3))   # [3, 6, 9, 12, 15, 18]
````

````python
def power(N, P):
    # TODO: base case P == 0; recursive case N * power(N, P - 1).
    pass

print(power(2, 5))
````

````python
# @title Solution { display-mode: "form" }
def power(N, P):
    if P == 0:          # Base case: N to the power 0 is 1
        return 1
    return N * power(N, P - 1)   # Recursive case

print(power(2, 5))   # 32
print(power(7, 0))   # 1
````

````python
# @title Solution { display-mode: "form" }
# Without recursion
def fib_loop(n):
    a, b = 0, 1
    for i in range(n):
        print(a, end=" ")
        a, b = b, a + b
    print()

# With recursion
def fib(k):
    if k == 0 or k == 1:    # Base cases
        return k
    return fib(k - 1) + fib(k - 2)

fib_loop(8)               # 0 1 1 2 3 5 8 13
for k in range(8):
    print(fib(k), end=" ")  # 0 1 1 2 3 5 8 13
print()
````

## SS5 — Positional arguments and *args practice

Bind the arguments to parameters before tracing the function body.

- Ordinary positional arguments fill parameters from left to right; write them before keyword arguments.
