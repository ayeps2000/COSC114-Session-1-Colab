```python
# @title Solution { display-mode: "form" }
mylist = ["cup", 20, 2.3, "football", True, "cup"]

def find_exact(items, target):
    for index, item in enumerate(items):
        if type(item) == type(target) and item == target:
            return index
    return -1

def find_ignore_case(words, target):
    for index, word in enumerate(words):
        if word.lower() == target.lower():
            return index
    return -1

print(find_exact(mylist, 1))
print(find_exact(mylist, True))
print(find_ignore_case(["Pen", "cup", "Book"], "CUP"))
assert find_exact(mylist, 1) == -1
assert find_exact(mylist, True) == 4
assert find_exact(mylist, 20) == 1
assert find_exact(mylist, "cup") == 0
assert find_exact([], 1) == -1
assert find_ignore_case(["Pen", "cup", "Book"], "CUP") == 1
assert find_ignore_case(["Pen"], "ink") == -1
```

```python
# @title Solution { display-mode: "form" }
def search_report(items, target):
    checks = 0
    for index, item in enumerate(items):
        checks += 1
        if item == target:
            return index, checks
    return -1, checks

print(search_report([20, 30, 2, 6, 99, 5], 6))
print(search_report([20, 30, 2, 6, 99, 5], 7))
for n in [10, 100, 1000]:
    data = list(range(n))
    index, checks = search_report(data, -1)
    print(n, "items:", checks, "checks for a missing value")

# Discussion: 'in' scans n items, then index() scans n items again: 2n checks.
# Store the index once instead:
#   position = find_first(items, x); if position != -1: ...
assert search_report([20, 30, 2, 6, 99, 5], 6) == (3, 4)
assert search_report([20, 30, 2, 6, 99, 5], 7) == (-1, 6)
assert search_report([], 1) == (-1, 0)
assert search_report(list(range(1000)), -1) == (-1, 1000)
```
