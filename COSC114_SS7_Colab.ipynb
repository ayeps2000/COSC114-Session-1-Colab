````python
mylist = ["cup", 20, 2.3, "football", True, "cup"]
mylist.extend(("tree", 24, 6.555))
print(mylist, len(mylist))
````

<div style="background:#f2f8f5;border-left:6px solid #198754;border-radius:8px;padding:10px 16px;"><p style="color:#198754;font-weight:700;margin:0 0 0.4em;">Answer</p><pre style="background:white;border:1px solid #d9e2ef;border-radius:6px;padding:8px 12px;font-family:Menlo,Consolas,monospace;font-size:18px;white-space:pre-wrap;">[&#x27;cup&#x27;, 20, 2.3, &#x27;football&#x27;, True, &#x27;cup&#x27;, &#x27;tree&#x27;, 24, 6.555] 9</pre><p style="margin:0;color:#526071;">extend() adds each item of the sequence separately; a tuple works as well as a list.</p></div>


````python
a = [1, 2]
a.append([3, 4])
print(a, len(a))
b = [1, 2]
b.extend([3, 4])
print(b, len(b))
````

<div style="background:#f2f8f5;border-left:6px solid #198754;border-radius:8px;padding:10px 16px;"><p style="color:#198754;font-weight:700;margin:0 0 0.4em;">Answer</p><pre style="background:white;border:1px solid #d9e2ef;border-radius:6px;padding:8px 12px;font-family:Menlo,Consolas,monospace;font-size:18px;white-space:pre-wrap;">[1, 2, [3, 4]] 3
[1, 2, 3, 4] 4</pre><p style="margin:0;color:#526071;">The slide's question: append() adds its argument as one item; extend() adds each item inside it.</p></div>


````python
m = [1]
m.extend("tree")
print(m)
````

````python
def inc(lis):
    for i in range(0, len(lis)):
        lis[i] += 1   # add 1 to each item in the list
    return lis

list1 = [1, 2, 3, 4, 5]
print("List before passing to the function:", list1)
inc(list1)
print("List after passing to the function:", list1)
````

````python
def inc(lis):
    newlist = lis.copy()
    for i in range(0, len(lis)):
        newlist[i] += 1
    return newlist

list1 = [1, 2, 3, 4, 5]
a = inc(list1)
print("Changed list after passing to the function:", a)
print("Original list after passing to the function:", list1)
````

````python
mylist = ["cup", 20, 2.3, "football", True, "cup"]

def linears(xlist, y):
    """This function performs linear search of y in list xlist
    and prints the value and the index"""
    c = 0
    for i in xlist:
        if i == y:
            print(y, "found at index", c)
            break
        c += 1
    else:
        print(y, "not found")

linears(mylist, 2.3)
linears(mylist, "tree")
````

````python
# @title Solution { display-mode: "form" }
temps = [23, 18, 25, 18, 30, 21, 18]

maxi = temps[0]
index = 0
for i in range(len(temps)):
    if temps[i] > maxi:
        maxi = temps[i]
        index = i
print(maxi, index)

count = 0
for t in temps:
    if t == 18:
        count += 1
print(count)

print([t for t in temps if t > 22])
print(sum(temps) / len(temps))

# The same answers with built-ins:
print(max(temps), temps.index(max(temps)), temps.count(18))
````

````python
items = ["cup", "pen", "cup", "book", "pen", "cup"]

# TODO: keep a list of the items already printed, so each one is printed once
# TODO: for each new item, count how many times it appears, then print the item and the count
````

````python
# @title Solution { display-mode: "form" }
items = ["cup", "pen", "cup", "book", "pen", "cup"]

seen = []
for item in items:
    if item not in seen:
        seen.append(item)
        count = 0
        for other in items:
            if other == item:
                count += 1
        print(item, count)
````

````python
a = [100, 50, 87, 876, 620, 3, 10]

# TODO: for each position start, find the largest item from start onwards
# TODO: swap it into position start
print(a)
````

````python
# @title Solution { display-mode: "form" }
a = [100, 50, 87, 876, 620, 3, 10]

for start in range(len(a)):
    maxi = a[start]
    index = start
    for i in range(start, len(a)):
        if a[i] > maxi:
            maxi = a[i]
            index = i
    a[start], a[index] = a[index], a[start]
print(a)
````
