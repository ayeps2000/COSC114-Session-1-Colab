# Running the code

- Run code cells one at a time, from top to bottom.
- Click a code cell and press `Shift` + `Enter` to run it and move on, or
  `Ctrl` + `Enter` to run it and stay put.
- When a cell calls `input()`, type the requested value in the text box below
  the cell and press `Enter`.
- Avoid **Runtime → Run all**, because cells using `input()` wait for you to type.

#### Applied: `**` and `input()` together

<div style="border-left:6px solid #2457a6;background:#eef3fa;border-radius:10px;padding:14px 18px;margin:6px 0;">
<p style="color:#2457a6 !important;font-weight:700;text-transform:uppercase;letter-spacing:0.02em;font-size:14px !important;margin:0 0 0.5em !important;">Two formulas, one operator</p>
<ul>
<li><code style="background:#ffffff;border-radius:4px;padding:0.05em 0.25em;">** 2</code> squares a value; <code style="background:#ffffff;border-radius:4px;padding:0.05em 0.25em;">** 0.5</code> takes its square root. No import needed.</li>
<li><code style="background:#ffffff;border-radius:4px;padding:0.05em 0.25em;">input()</code> hands back text, so <code style="background:#ffffff;border-radius:4px;padding:0.05em 0.25em;">float()</code> has to come first or the arithmetic fails.</li>
<li><code style="background:#ffffff;border-radius:4px;padding:0.05em 0.25em;">**</code> binds tighter than <code style="background:#ffffff;border-radius:4px;padding:0.05em 0.25em;">*</code>, so <code style="background:#ffffff;border-radius:4px;padding:0.05em 0.25em;">0.5 * m * v ** 2</code> squares <code style="background:#ffffff;border-radius:4px;padding:0.05em 0.25em;">v</code> before multiplying.</li>
</ul>
<p style="margin:0 !important;font-size:15px !important;color:#526071 !important;">Try side 3 and side 4, then mass 4 kg at 3 m/s.</p>
</div>

```python
# Pythagorean Theorem Calculator
# Demonstrates input, processing, and output


# Get input from user
a = float(input("Enter the length of side a: "))
b = float(input("Enter the length of side b: "))

# Calculate hypotenuse: c = (a**2 + b**2) raised to the power 1/2
c = (a**2 + b**2) ** 0.5

# Display result
print("Side a:", a)
print("Side b:", b)
print("Hypotenuse c:", c)
```

```python
# Kinetic energy of a moving mass: KE = 0.5 * m * v**2

mass_kg = float(input("Mass in kg: "))
speed_ms = float(input("Speed in m/s: "))

# ** runs before *, so the speed is squared first
energy_j = 0.5 * mass_kg * speed_ms ** 2

print("Mass:", mass_kg, "kg")
print("Speed:", speed_ms, "m/s")
print("Kinetic energy:", energy_j, "J")
```

```python
a = input("Enter the value of a: ")     # <-- no int() !
b = input("Enter the value of b: ")     # <-- no int() !

if a > b:
    print("a is larger")
else:
    print("b is larger")
```

```python
# The same buggy program, run over all three pairs at once.
for a, b in [("9", "10"), ("100", "25"), ("7", "3")]:
    if a > b:                       # text comparison, not number comparison
        says = "a is larger"
    else:
        says = "b is larger"

    truth = "a is larger" if int(a) > int(b) else "b is larger"
    verdict = "right" if says == truth else "WRONG"
    print(f"{a:>4} and {b:>4}  ->  program says {says:<12} truth: {truth:<12} {verdict}")
```

```python
# The fix is one function wide. Convert at the door, before any comparison.
a = int(input("Enter the value of a: "))
b = int(input("Enter the value of b: "))

if a > b:
    print("a is larger")
else:
    print("b is larger")
```

```python
temp = int(input("Temperature: "))

if temp >= 25:
    print("Hot day")

if temp <= 25:              # <-- independent if: overlaps at exactly 25
    print("Cool day")
```

```python
# The same two-if program, run over all three temperatures.
for temp in [30, 10, 25]:
    print(temp, "->", end=" ")

    if temp >= 25:
        print("Hot day", end=" ")

    if temp <= 25:
        print("Cool day", end=" ")

    print()
```

```python
# The fix: one question, one door. The else needs no condition of its own.
for temp in [30, 10, 25]:
    if temp >= 25:
        print(temp, "-> Hot day")
    else:
        print(temp, "-> Cool day")
```

#### Applied: leaving a loop early &mdash; `break` and `for ... else`

A message that fails to send is retried three times. `break` walks out the moment it succeeds; the `else` on the `for` runs only if the loop was never broken out of.


```python
# Three attempts. number counts 0, 1, 2 — so number + 1 reads as 1, 2, 3.
for number in range(3):
    print("Attempt", number + 1, "." * (number + 1))
```

```python
# break — stop the loop the moment the message goes through
successful = True

for number in range(3):
    print("Attempt", number + 1)
    if successful:
        print("Successful")
        break                      # no attempt 2 or 3
```

```python
# for ... else — the else belongs to the FOR, not to the if.
# It runs only when the loop finishes without ever hitting break.
successful = False

for number in range(3):
    print("Attempt", number + 1)
    if successful:
        print("Successful")
        break
else:
    print("Attempted 3 times and failed")
```

#### Applied: nested loops

A loop inside a loop. The inner loop runs to completion on *every* trip of the outer one, so this prints 5 &times; 3 = 15 lines.


```python
# Outer loop: x = 0, 1, 2, 3, 4
# Inner loop: for each x, y runs 0, 1, 2 all over again
for x in range(5):
    for y in range(3):
        print(f"({x}, {y})")
```

#### Applied: what `range()` returns, and what else a `for` can walk

`range()` does not hand back a list &mdash; it hands back a `range` object. What makes it usable in a `for` is that it is *iterable*, and strings and lists are iterable too.


```python
print(type(5))            # <class 'int'>      — a primitive type
print(type(range(5)))     # <class 'range'>    — a complex type
```

```python
# A string is iterable: one character per trip
for x in "Python":
    print(x)
```

```python
# So is a list: one item per trip
for x in ["Mosh", "John", "Sarah"]:
    print(x)
```

#### Applied: count the even numbers between 1 and 10

Walk every number, test it with `%`, print the even ones and keep a running count. No `step` argument &mdash; the `if` does the filtering.


```python
# Print the even numbers between 1 and 10, then say how many there were
count = 0

for number in range(1, 10):
    if number % 2 == 0:            # remainder 0 means even
        print(number)
        count += 1
            
print(f"We have {count} even numbers")
```
