````python
# Customer 1: Alice orders an iced latte ($4.50)
price_alice = 4.50
tax_alice = price_alice * 0.05
total_alice = price_alice + tax_alice + 2.00
print(f"Alice's Total: ${total_alice:.2f}")

# Customer 2: Bob orders a cold brew and a muffin ($7.80)
price_bob = 7.80
tax_bob = price_bob * 0.05
total_bob = price_bob + tax_bob + 2.00
print(f"Bob's Total: ${total_bob:.2f}")

# Customer 3: Charlie orders an espresso ($3.00)
price_charlie = 3.00
tax_charlie = price_charlie * 0.05
total_charlie = price_charlie + tax_charlie + 2.00
print(f"Charlie's Total: ${total_charlie:.2f}")
````

````python
def calculate_order_total(subtotal, customer_name):
    tax = subtotal * 0.05
    delivery_fee = 2.00
    total = subtotal + tax + delivery_fee
    return f"{customer_name}'s Total: ${total:.2f}"

# Now, running an order is just one clean call:
print(calculate_order_total(4.50, "Alice"))
print(calculate_order_total(7.80, "Bob"))
print(calculate_order_total(3.00, "Charlie"))
````

Both cells print the same three totals. To change the tax or the delivery fee, the first version needs
three edits (one per customer); the function needs one.


#### Slide 3 &middot; live code below

<div style="background:#ffffff;border:1px solid #d9e2ef;border-radius:10px;padding:6px;margin:2px 0 4px;text-align:center;"><img alt="PowerPoint slide 3: Functions: the monster arena" loading="lazy" decoding="async" src="https://raw.githubusercontent.com/ayeps2000/COSC114-Session-1-Colab/main/slides/ss5/slide-3.png" style="display:block;margin:0 auto;width:auto;max-width:100%;max-height:62vh;"></div>

````python
# --- BATTLE 1: Goblin (45 HP, low defense) ---
print("=== A wild Goblin appears! ===")
goblin_hp = 45
goblin_turns = 0

while goblin_hp > 0:
    goblin_turns += 1
    # Roll a simulated attack value between 1 and 20
    roll = (goblin_turns * 7) % 20 + 1

    if roll >= 18:
        damage = 25
        print(f"Turn {goblin_turns}: CRITICAL HIT! Dealt {damage} damage.")
    elif roll <= 5:
        damage = 5
        print(f"Turn {goblin_turns}: Glancing blow... Dealt {damage} damage.")
    else:
        damage = 12
        print(f"Turn {goblin_turns}: Solid strike. Dealt {damage} damage.")

    goblin_hp -= damage
    if goblin_hp < 0:
        goblin_hp = 0
    print(f"Goblin HP remaining: {goblin_hp}")

print(f"Goblin defeated in {goblin_turns} turns!\n")

# --- BATTLE 2: Stone Golem (80 HP, high defense) ---
print("=== A massive Stone Golem appears! ===")
golem_hp = 80
golem_turns = 0

while golem_hp > 0:
    golem_turns += 1
    roll = (golem_turns * 7) % 20 + 1

    if roll >= 18:
        damage = 25
        print(f"Turn {golem_turns}: CRITICAL HIT! Dealt {damage} damage.")
    elif roll <= 5:
        damage = 5
        print(f"Turn {golem_turns}: Glancing blow... Dealt {damage} damage.")
    else:
        damage = 12
        print(f"Turn {golem_turns}: Solid strike. Dealt {damage} damage.")

    golem_hp -= damage
    if golem_hp < 0:
        golem_hp = 0
    print(f"Golem HP remaining: {golem_hp}")

print(f"Stone Golem defeated in {golem_turns} turns!\n")
````

````python
def fight_monster(monster_name, max_hp):
    """Simulates combat against a monster and returns total turns taken."""
    print(f"=== A wild {monster_name} appears! ===")
    hp = max_hp
    turns = 0

    while hp > 0:
        turns += 1
        roll = (turns * 7) % 20 + 1  # Deterministic roll simulation

        # Combat decision logic
        if roll >= 18:
            damage = 30
            print(f"Turn {turns}: CRITICAL HIT! Dealt {damage} damage.")
        elif roll <= 5:
            damage = 5
            print(f"Turn {turns}: Glancing blow... Dealt {damage} damage.")
        else:
            damage = 14
            print(f"Turn {turns}: Solid strike. Dealt {damage} damage.")

        hp -= damage
        if hp < 0:
            hp = 0
        print(f"{monster_name} HP remaining: {hp}")

    print(f"Victory! {monster_name} defeated in {turns} turns.\n")
    return turns

# --- Main Game Loop: Clean, readable, and easy to scale ---
roster = [
    ("Goblin", 45),
    ("Stone Golem", 80),
    ("Dragon", 120),
]

total_arena_turns = 0

for name, health in roster:
    turns_taken = fight_monster(name, health)
    total_arena_turns += turns_taken

print(f"Arena cleared! Total turns across all battles: {total_arena_turns}")
````

The function holds the `while` loop and the `if`/`elif`/`else` checks once, and **returns** the number of turns
so the main loop can add them up. Another monster is one more line in `roster`.

**When `n <= 0`:** `range(1, n + 1)` is empty, so the loop never assigns `i` and
`avg = s / i` raises `UnboundLocalError: cannot access local variable 'i' where it is not associated with a value`.
Returning early for `n <= 0` and dividing by `n` avoids both problems:


````python
def natural(n):
    if n <= 0:
        print("n must be a positive whole number")
        return
    s = 0
    for i in range(1, n + 1):
        s = s + i
    avg = s / n
    print("The sum of the first", n, "natural numbers is", s)
    print("The average of first", n, "natural numbers is", avg)

natural(5)
natural(0)
````

**Warm-up — tracing `mystery(multiplier, *args)`.**


````python
def mystery(multiplier, *args):
    result = []
    for item in args:
        result.append(item * multiplier)
    return result

print(mystery(3, 1, 2, 4))
print(mystery(2, "a", "b"))
print(mystery(5))
````

**Trace table: one column per call.**

| | Call 1 | Call 2 | Call 3 |
|---|---|---|---|
| **Call** | `mystery(3, 1, 2, 4)` | `mystery(2, 'a', 'b')` | `mystery(5)` |
| **`multiplier`** | `3` | `2` | `5` |
| **`args`** | `(1, 2, 4)` | `('a', 'b')` | `()` |
| **Loop body** | `1 * 3` → `3`, `2 * 3` → `6`, `4 * 3` → `12` | `'a' * 2` → `'aa'`, `'b' * 2` → `'bb'` | loop never runs |
| **Returns** | `[3, 6, 12]` | `['aa', 'bb']` | `[]` |


**Exam-style example (Midterm Fall 2025, Q4.3).** What does the program print?

(A) 4 (B) 6 (C) 7 (D) 9 (E) 14 (F) 19 (G) None of the above


````python
def mystery(multiplier, *args):
    total = 0
    count = 0
    for n in args:
        if n % 2 == 0:
            total += n * multiplier
        else:
            total -= n
        count += 1
        if count == multiplier:
            total += 1
    return total

print(mystery(3, 2, 5, 4, 7))
````

#### Slide 54C — 1. Universal Mathematical Formula for Recursion

<div class="cosc-practice-slide">

<div style="display:grid;grid-template-columns:repeat(auto-fit,minmax(300px,1fr));gap:16px;align-items:stretch;margin:8px 0 12px;">
<div style="background:#f8fafd;border:1px solid #d9e2ef;border-radius:10px;padding:12px 16px;">

**Mathematical Formula**

$$
f(n)=
\begin{cases}
b, & n=0 \\
a\cdot f(n-1)+c, & n>0
\end{cases}
$$

</div>
<div style="background:#f8fafd;border:1px solid #d9e2ef;border-radius:10px;padding:12px 16px;">

**Equivalent Python Template**

```python
def recursive(n):
    if n == 0:
        return b

    return a * recursive(n - 1) + c
```

</div>
</div>

Here, `a`, `b`, and `c` must be defined before calling the function.

| Symbol | Meaning |
|---|---|
| `a` | Multiplication factor |
| `b` | Base case value |
| `c` | Value added at each recursive step |
| `n` | Current input |

</div>


````python
a = 2   # Multiplication factor
b = 2   # Base case value: f(0) = 2
c = 3   # Value added at each recursive step

def recursive(n):
    if n == 0:
        return b

    return a * recursive(n - 1) + c

print("f(0) =", recursive(0))
print("f(1) =", recursive(1))
print("f(2) =", recursive(2))
print("f(3) =", recursive(3))
````

#### Slide 54D — 2. General Formula (Factorial, Natural Sum, etc.)

<div class="cosc-practice-slide">

<div style="display:grid;grid-template-columns:repeat(auto-fit,minmax(300px,1fr));gap:16px;align-items:stretch;margin:8px 0 12px;">
<div style="background:#f8fafd;border:1px solid #d9e2ef;border-radius:10px;padding:12px 16px;">

**Mathematical Formula**

$$
f(n)=
\begin{cases}
b, & n=0 \\
a(n)\cdot f(n-1)+c(n), & n>0
\end{cases}
$$

</div>
<div style="background:#f8fafd;border:1px solid #d9e2ef;border-radius:10px;padding:12px 16px;">

**Equivalent Python Template**

```python
def recursive(n):
    if n == 0:
        return b

    return a(n) * recursive(n - 1) + c(n)
```

</div>
</div>

Here, `a(n)` and `c(n)` are functions of `n` rather than fixed constants, and must be defined appropriately:
- **Factorial ($n!$):** take `b = 1`, `a(n) = n`, and `c(n) = 0` $\rightarrow$ `return n * recursive(n - 1)`
- **Natural sum ($1 + 2 + \dots + n$):** take `b = 0`, `a(n) = 1`, and `c(n) = n` $\rightarrow$ `return recursive(n - 1) + n`

</div>


````python
# Factorial using the general template: b = 1, a(n) = n, c(n) = 0
b = 1
a = lambda n: n
c = lambda n: 0

def recursive(n):
    if n == 0:
        return b

    return a(n) * recursive(n - 1) + c(n)

print("Factorial f(4) =", recursive(4))
````

#### Slide 54E — 3. Universal Formula for Two Recursive Calls

<div class="cosc-practice-slide">

<div style="display:grid;grid-template-columns:repeat(auto-fit,minmax(300px,1fr));gap:16px;align-items:stretch;margin:8px 0 12px;">
<div style="background:#f8fafd;border:1px solid #d9e2ef;border-radius:10px;padding:12px 16px;">

**Mathematical Formula**

$$
f(n)=
\begin{cases}
B(n), & n\le 1 \\
G(n,f(n-1),f(n-2)), & n>1
\end{cases}
$$

</div>
<div style="background:#f8fafd;border:1px solid #d9e2ef;border-radius:10px;padding:12px 16px;">

**Equivalent Python Template**

```python
def recursive(n):
    if n <= 1:
        return B(n)

    return G(
        n,
        recursive(n - 1),
        recursive(n - 2)
    )
```

</div>
</div>

- `B(n)` specifies the base case values ($n \le 1$).
- `G(...)` combines the current input with the two recursive results. Both `B` and `G` must be defined for the template to run.

</div>


````python
# Fibonacci using the two-call template: B(n) = n, G(n, r1, r2) = r1 + r2
B = lambda n: n
G = lambda n, r1, r2: r1 + r2

def recursive(n):
    if n <= 1:
        return B(n)

    return G(
        n,
        recursive(n - 1),
        recursive(n - 2)
    )

print([recursive(k) for k in range(7)])
````

#### Slide 54F — 4. Examples: Mathematical Formula $\rightarrow$ Python

<div class="cosc-practice-slide">

| Problem | Mathematical recurrence | Python recursive statement | Base case |
|---|---|---|---|
| Factorial | $f(n)=n f(n-1)$ | `return n * f(n - 1)` | $f(0)=1$ |
| Natural sum | $f(n)=n+f(n-1)$ | `return n + f(n - 1)` | $f(0)=0$ |
| Powers of 3 | $f(n)=3f(n-1)$ | `return 3 * f(n - 1)` | $f(0)=1$ |
| Multiply by 2, add 3 | $f(n)=2f(n-1)+3$ | `return 2 * f(n - 1) + 3` | Example: $f(0)=2$ |
| Fibonacci | $f(n)=f(n-1)+f(n-2)$ | `return f(n - 1) + f(n - 2)` | $f(0)=0,\ f(1)=1$ |

**Important:** These are templates for particular families of recurrences, not formulas that represent every possible recursive algorithm. Recursion always needs a **termination condition** and a **smaller subproblem** (or other argument progressing toward termination).

</div>


````python
def fact(n):
    return 1 if n == 0 else n * fact(n - 1)

def natural_sum(n):
    return 0 if n == 0 else n + natural_sum(n - 1)

def power_of_3(n):
    return 1 if n == 0 else 3 * power_of_3(n - 1)

def mult2_add3(n):
    return 2 if n == 0 else 2 * mult2_add3(n - 1) + 3

def fib(n):
    return n if n <= 1 else fib(n - 1) + fib(n - 2)

print("fact(4) =", fact(4))
print("natural_sum(4) =", natural_sum(4))
print("power_of_3(4) =", power_of_3(4))
print("mult2_add3(3) =", mult2_add3(3))
print("fib(6) =", fib(6))
````

#### Slide 57A — Functions that return functions (closures)

<div class="cosc-practice-slide">
<p><code>mul = lambda x: (lambda y: y * x)</code> defines a function inside another function:</p>
<pre><code>def mul(x):
    def multiply(y):
        return y * x
    return multiply</code></pre>
<table>
<tr><th>Call</th><th>Effect</th></tr>
<tr><td><code>times4 = mul(4)</code></td><td>Binds <code>x = 4</code> and returns the inner function</td></tr>
<tr><td><code>times4(2)</code></td><td>Passes <code>y = 2</code> into the returned function and returns <code>2 * 4 = 8</code></td></tr>
</table>
<p>An inner function that remembers variables from its enclosing scope (<code>x = 4</code>) is a <strong>closure</strong>.</p>
</div>


````python
def mul(x):
    def multiply(y):
        return y * x

    return multiply

times4 = mul(4)

print(times4(2))
````

#### Slide 57B — Closures and chained calls (currying)

<div class="cosc-practice-slide">
<ul>
<li>Each call to <code>mul(x)</code> creates a function with its own saved value of <code>x</code> (<code>times4</code> keeps <code>4</code>; <code>times10</code> keeps <code>10</code>).</li>
<li>The returned function can also be called immediately in one expression (<strong>currying</strong>):</li>
</ul>
<pre><code>mul(4)       # Returns the inner function (y -> y * 4)
mul(4)(2)    # Calls that function with y = 2 -> 8</code></pre>
</div>


````python
mul = lambda x: (lambda y: y * x)

times4 = mul(4)
times10 = mul(10)

print(times4(3))
print(times10(3))
````

30</pre><p style="margin:0;color:#526071;">times4 retains x = 4 (3 * 4 = 12) and times10 retains x = 10 (3 * 10 = 30).</p></div>

````python
# @title Solution { display-mode: "form" }
# Without recursion
def fib_loop(n):
    a, b = 0, 1
    for i in range(n):
        print(a, end=" ")
        a, b = b, a + b
    print()

# With recursion
def fib(k):
    if k == 0 or k == 1:    # Base cases
        return k
    return fib(k - 1) + fib(k - 2)

fib_loop(8)               # 0 1 1 2 3 5 8 13
for k in range(8):
    print(fib(k), end=" ")  # 0 1 1 2 3 5 8 13
print()
````

#### Slide 71B — Packing with *args and **kwargs

<div class="cosc-practice-slide">
<ul>
<li><strong><code>*args</code> (variable positional arguments):</strong> packs any number of positional arguments into a <code>tuple</code>.</li>
<li><strong><code>**kwargs</code> (variable keyword arguments):</strong> packs any number of keyword arguments into a <code>dict</code>.</li>
</ul>
</div>


````python
def sum_all(*args):
    print(f"args type: {type(args)}, contents: {args}")
    return sum(args)

result = sum_all(10, 20, 30, 40)
print("result =", result)
````

result = 100</pre><p style="margin:0;color:#526071;">The four positional arguments are packed into one tuple, args. sum(args) adds them: 10 + 20 + 30 + 40 = 100.</p></div>

#### Slide 71C — Parameter ordering with *args and **kwargs

<div class="cosc-practice-slide">
<p>When combining parameter types in a function definition, the order is:</p>
<ol>
<li>Standard positional parameters</li>
<li><code>*args</code></li>
<li>Keyword-only parameters (with or without defaults)</li>
<li><code>**kwargs</code></li>
</ol>
<pre><code>def func(name, *activities, age=18, **locations):</code></pre>
</div>


````python
def make_call(protocol, *args, **kwargs):
    print(f"Protocol: {protocol}")
    print(f"Extra positional (tuple): {args}")
    print(f"Extra keyword (dict): {kwargs}")

make_call("HTTPS", "api.example.com", 443, timeout=30, retries=3)
````

#### Slide 71D — Unpacking arguments in a function call

<div class="cosc-practice-slide">
<p>In a function <strong>call</strong>, <code>*</code> and <code>**</code> unpack a collection into separate arguments:</p>
<ul>
<li><code>add(*numbers)</code> unpacks a sequence into positional arguments.</li>
<li><code>add(**data)</code> unpacks a dictionary into keyword arguments (keys must match parameter names).</li>
</ul>
<p>The names <code>args</code> and <code>kwargs</code> are conventions; the <code>*</code> and <code>**</code> operators perform the packing and unpacking.</p>
</div>


````python
def add(a, b, c):
    return a + b + c

numbers = [1, 2, 3]
print(add(*numbers))

data = {"a": 4, "b": 5, "c": 6}
print(add(**data))
````
