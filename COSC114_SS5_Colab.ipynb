````python
# Customer 1: Alice orders an iced latte ($4.50)
price_alice = 4.50
tax_alice = price_alice * 0.05
total_alice = price_alice + tax_alice + 2.00
print(f"Alice's Total: ${total_alice:.2f}")

# Customer 2: Bob orders a cold brew and a muffin ($7.80)
price_bob = 7.80
tax_bob = price_bob * 0.05
total_bob = price_bob + tax_bob + 2.00
print(f"Bob's Total: ${total_bob:.2f}")

# Customer 3: Charlie orders an espresso ($3.00)
price_charlie = 3.00
tax_charlie = price_charlie * 0.05
total_charlie = price_charlie + tax_charlie + 2.00
print(f"Charlie's Total: ${total_charlie:.2f}")
````

````python
def calculate_order_total(subtotal, customer_name):
    tax = subtotal * 0.05
    delivery_fee = 2.00
    total = subtotal + tax + delivery_fee
    return f"{customer_name}'s Total: ${total:.2f}"

# Now, running an order is just one clean call:
print(calculate_order_total(4.50, "Alice"))
print(calculate_order_total(7.80, "Bob"))
print(calculate_order_total(3.00, "Charlie"))
````

Both cells print the same three totals. To change the tax or the delivery fee, the first version needs
three edits (one per customer); the function needs one.


#### Slide 3 &middot; live code below

<div style="background:#ffffff;border:1px solid #d9e2ef;border-radius:10px;padding:6px;margin:2px 0 4px;text-align:center;"><img alt="PowerPoint slide 3: Functions: the monster arena" loading="lazy" decoding="async" src="https://raw.githubusercontent.com/ayeps2000/COSC114-Session-1-Colab/main/slides/ss5/slide-3.png" style="display:block;margin:0 auto;width:auto;max-width:100%;max-height:62vh;"></div>

````python
# --- BATTLE 1: Goblin (45 HP, low defense) ---
print("=== A wild Goblin appears! ===")
goblin_hp = 45
goblin_turns = 0

while goblin_hp > 0:
    goblin_turns += 1
    # Roll a simulated attack value between 1 and 20
    roll = (goblin_turns * 7) % 20 + 1

    if roll >= 18:
        damage = 25
        print(f"Turn {goblin_turns}: CRITICAL HIT! Dealt {damage} damage.")
    elif roll <= 5:
        damage = 5
        print(f"Turn {goblin_turns}: Glancing blow... Dealt {damage} damage.")
    else:
        damage = 12
        print(f"Turn {goblin_turns}: Solid strike. Dealt {damage} damage.")

    goblin_hp -= damage
    if goblin_hp < 0:
        goblin_hp = 0
    print(f"Goblin HP remaining: {goblin_hp}")

print(f"Goblin defeated in {goblin_turns} turns!\n")

# --- BATTLE 2: Stone Golem (80 HP, high defense) ---
print("=== A massive Stone Golem appears! ===")
golem_hp = 80
golem_turns = 0

while golem_hp > 0:
    golem_turns += 1
    roll = (golem_turns * 7) % 20 + 1

    if roll >= 18:
        damage = 25
        print(f"Turn {golem_turns}: CRITICAL HIT! Dealt {damage} damage.")
    elif roll <= 5:
        damage = 5
        print(f"Turn {golem_turns}: Glancing blow... Dealt {damage} damage.")
    else:
        damage = 12
        print(f"Turn {golem_turns}: Solid strike. Dealt {damage} damage.")

    golem_hp -= damage
    if golem_hp < 0:
        golem_hp = 0
    print(f"Golem HP remaining: {golem_hp}")

print(f"Stone Golem defeated in {golem_turns} turns!\n")
````

````python
def fight_monster(monster_name, max_hp):
    """Simulates combat against a monster and returns total turns taken."""
    print(f"=== A wild {monster_name} appears! ===")
    hp = max_hp
    turns = 0

    while hp > 0:
        turns += 1
        roll = (turns * 7) % 20 + 1  # Deterministic roll simulation

        # Combat decision logic
        if roll >= 18:
            damage = 30
            print(f"Turn {turns}: CRITICAL HIT! Dealt {damage} damage.")
        elif roll <= 5:
            damage = 5
            print(f"Turn {turns}: Glancing blow... Dealt {damage} damage.")
        else:
            damage = 14
            print(f"Turn {turns}: Solid strike. Dealt {damage} damage.")

        hp -= damage
        if hp < 0:
            hp = 0
        print(f"{monster_name} HP remaining: {hp}")

    print(f"Victory! {monster_name} defeated in {turns} turns.\n")
    return turns

# --- Main Game Loop: Clean, readable, and easy to scale ---
roster = [
    ("Goblin", 45),
    ("Stone Golem", 80),
    ("Dragon", 120),
]

total_arena_turns = 0

for name, health in roster:
    turns_taken = fight_monster(name, health)
    total_arena_turns += turns_taken

print(f"Arena cleared! Total turns across all battles: {total_arena_turns}")
````

The function holds the `while` loop and the `if`/`elif`/`else` checks once, and **returns** the number of turns
so the main loop can add them up. Another monster is one more line in `roster`.

**What if n was 0 or negative?** `range(1, n + 1)` is empty, so the loop never assigns `i` and
`avg = s / i` fails with `UnboundLocalError: cannot access local variable 'i' where it is not associated with a value`.
Divide by `n` instead of `i`, and reject `n <= 0` before doing any work:


````python
def natural(n):
    if n <= 0:
        print("n must be a positive whole number")
        return
    s = 0
    for i in range(1, n + 1):
        s = s + i
    avg = s / n
    print("The sum of the first", n, "natural numbers is", s)
    print("The average of first", n, "natural numbers is", avg)

natural(5)
natural(0)
````

**Trace table: one column per call.**

| | Call 1 | Call 2 | Call 3 |
|---|---|---|---|
| **Call** | `mystery(3, 1, 2, 4)` | `mystery(2, 'a', 'b')` | `mystery(5)` |
| **`multiplier`** | `3` | `2` | `5` |
| **`args`** | `(1, 2, 4)` | `('a', 'b')` | `()` |
| **Loop body** | `1 * 3` → `3`, `2 * 3` → `6`, `4 * 3` → `12` | `'a' * 2` → `'aa'`, `'b' * 2` → `'bb'` | loop never runs |
| **Returns** | `[3, 6, 12]` | `['aa', 'bb']` | `[]` |


**Exam-style example (Midterm Fall 2025, Q4.3).** What does the program print?

(A) 4 (B) 6 (C) 7 (D) 9 (E) 14 (F) 19 (G) None of the above


````python
def mystery(multiplier, *args):
    total = 0
    count = 0
    for n in args:
        if n % 2 == 0:
            total += n * multiplier
        else:
            total -= n
        count += 1
        if count == multiplier:
            total += 1
    return total

print(mystery(3, 2, 5, 4, 7))
````

````python
def fact(n):
    print("call fact(" + str(n) + ")")
    if n == 1 or n == 0:
        print("fact(" + str(n) + ") returns 1")
        return 1
    result = n * fact(n - 1)
    print("fact(" + str(n) + ") returns", result)
    return result

print(fact(3))
````

````python
# @title Solution { display-mode: "form" }
# Without recursion
def fib_loop(n):
    a, b = 0, 1
    for i in range(n):
        print(a, end=" ")
        a, b = b, a + b
    print()

# With recursion
def fib(k):
    if k == 0 or k == 1:    # Base cases
        return k
    return fib(k - 1) + fib(k - 2)

fib_loop(8)               # 0 1 1 2 3 5 8 13
for k in range(8):
    print(fib(k), end=" ")  # 0 1 1 2 3 5 8 13
print()
````

<div style="background:#ffffff;border:1px solid #d9e2ef;border-radius:10px;padding:6px;margin:2px 0 4px;text-align:center;"><img alt="PowerPoint slide 71: Combining *args and **kwargs" loading="lazy" decoding="async" src="https://raw.githubusercontent.com/ayeps2000/COSC114-Session-1-Colab/main/slides/ss5/slide-71.png" style="display:block;margin:0 auto;width:auto;max-width:100%;max-height:62vh;"></div>

````python
def func(name, *activities, age=18, **locations):
    print(f"{name} (Age: {age})")
    print("Favorite Activities:")
    for activity in activities:
        print(activity)
    print("Has traveled to:")
    for city in locations.values():
        print(city)

# Function Call
func("Fulan", "Reading", "Traveling", age=22, city1="AD", city2="RAK")
````
