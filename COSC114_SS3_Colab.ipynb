````python
# A function called by itself vs a method called ON a value
name = 'John'

# function: the value goes inside the ()
print("len(name)      :", len(name))

# method: the value comes before the dot
print("name.isalpha() :", name.isalpha())

# "What kind of thing is it?" decides which tools it carries
print("type(name)     :", type(name))      # <class 'str'>
print("type(20)       :", type(20))        # <class 'int'>

# The toolkit belongs to the kind of value, not the variable name
print("'20'.isdigit() :", '20'.isdigit())  # text, so str tools apply
````

````python
# Ask a number for a text tool: no such tool exists
print("'20'.isdigit() :", '20'.isdigit())   # True, '20' is text
print("type of '20'   :", type('20'))       # str: text has the is... tools
print("type of 20     :", type(20))         # int: a number has none of them

# Remove the # on the next line in class to watch Python refuse it:
# print((20).isdigit())
# AttributeError: 'int' object has no attribute 'isdigit'
#   'int' object -> the value
#   attribute    -> anything reached through the dot

# Without the () you get the tool itself, not the answer:
print("isalpha   :", 'John'.isalpha)        # the method object
print("isalpha() :", 'John'.isalpha())      # the answer we wanted
````

````python
# String inspection methods from Slide 3
name = 'John'
print("name.isalpha()  :", name.isalpha())     # True (only letters)

age = '20'
print("age.isdigit()   :", age.isdigit())      # True (all characters are digits)

print("'john'.islower() :", 'john'.islower())   # True (all lowercase)
print("'john1'.isalpha():", 'john1'.isalpha()) # False (contains digit '1')

# Whitespace and case inspection:
print("'  \\t\\n'.isspace() :", '  \t\n'.isspace())   # True (whitespace characters only)
print("'PYTHON'.isupper() :", 'PYTHON'.isupper())  # True (all uppercase)
````

#### Practice 3A · Tier 1: Clean and compare

Use `.strip()` when your input rules allow surrounding whitespace.
It returns a cleaned copy; `.lower()` and `.upper()` return new copies too.

| Tool | Example | Result |
| --- | --- | --- |
| `.strip()` | `' 20 '.strip()` | `'20'` |
| `.lower()` | `'YES'.lower()` | `'yes'` |
| `.upper()` | `'yes'.upper()` | `'YES'` |
| `.startswith()` | `'quit'.startswith('q')` | `True` |
| `in` operator | `'a' in 'cat'` | `True` |

**Choose the match:** `==` checks the whole value, `.startswith()` checks a
prefix, and `in` checks anywhere. Slide 29 stops on exactly `'a'`.
Check a space sentinel (slide 28) **before** stripping it away.

<span class="cosc-practice-slide"></span>


````python
# Spaces affect digit checks; cleaning is a separate step.
raw = ' 20 '
print(raw.isdigit(), raw.strip().isdigit())
print(repr(raw))                 # the original value is unchanged

reply = 'Yes '
print(reply.strip().lower() == 'yes')
print(reply.strip().upper())

# Prefix, membership, and exact equality answer different questions.
print('quit'.startswith('q'))
print('a' in 'cat')
print('cat' == 'a')               # False: do not stop on this value
````

````python
# Predict the output
reply = "  YES  "
print(reply.isalpha(), reply.strip().isalpha())
print(reply.strip().lower() == "yes")
print("yellow".startswith("y"), "yellow" == "yes")
````

#### Practice 3B · Tier 2: Six more string checks

| Method | Example | Result |
| --- | --- | --- |
| `.istitle()` | `'John Smith'.istitle()` | `True` |
| `.isdecimal()` | `'²'.isdecimal()` | `False` |
| `.isnumeric()` | `'½'.isnumeric()` | `True` |
| `.isidentifier()` | `'my_var'.isidentifier()` | `True` |
| `.isascii()` | `'Ali'.isascii()` | `True` |
| `.isprintable()` | `' \n'.isprintable()` | `False` |

Decimal digits are a subset of digit characters, which are a subset of numeric
characters. **Decimal does not mean ASCII-only:** `'٢٠'.isdecimal()` is `True`
and `int('٢٠')` is `20`. But `int('²')` fails even though `'²'.isdigit()` is `True`.

`'for'.isidentifier()` is `True`; it is still a keyword, so it cannot be a variable name.

<span class="cosc-practice-slide"></span>


````python
# Run every method in the table, then compare the counterexamples.
print('John Smith'.istitle(), 'John Smith'.isalpha())
print('²'.isdecimal(), '²'.isdigit(), '²'.isnumeric())
print('½'.isnumeric(), '½'.isdigit())
print('my_var'.isidentifier(), '123'.isidentifier())
print('for'.isidentifier())       # spelling is valid; the keyword is reserved
print('Ali'.isascii(), '٢٠'.isascii())
print(' \n'.isprintable(), 'a b'.isprintable())
print('٢٠'.isdecimal(), '٢٠'.isdigit(), int('٢٠'))
````

#### Practice 3C · Tier 3: Three input traps

`.isdigit()` requires **at least one character**, with every character a digit.
It does not mean “can this text be converted to a number?”

| Input | `.isdigit()` | Reason |
| --- | --- | --- |
| `''` | `False` | No characters |
| `'-5'` | `False` | The minus sign is not a digit |
| `'3.14'` | `False` | The decimal point is not a digit |

**Empty-string exceptions:** `''.isascii()` and `''.isprintable()` are both `True`.
The six inspection methods on source slide 3 return `False` for `''`.

**Practice policy:** accept 1–12 Unicode decimal digits with an optional leading
`+` or `-`, after trimming outside whitespace. Reject decimal points and internal
spaces. In the code, `digits[1:]` removes the first character after checking a sign.

<span class="cosc-practice-slide"></span>


````python
for text in ['', '-5', '3.14', '42']:
    print(repr(text), text.isdigit())

# Do not generalize the empty-string rule to every is... method.
print('empty isascii:', ''.isascii())
print('empty isprintable:', ''.isprintable())

# Conversions accept syntax that isdigit() rejects.
print(int('-5'))
print(float('3.14'))
````

````python
# Predict the output
print(''.isdigit(), ''.isascii(), ''.isprintable())
print('-5'.isdigit(), int('-5'))
print('3.14'.isdigit(), float('3.14'))
````

````python
samples = ['42', ' 42 ', '-5', '+5', '٢٠', '', '   ',
           'abc', '3.14', '-', '--5', '4 2', '1234567890123']
# TODO: loop over samples and apply the policy on the practice slide.
# Hint: after detecting a leading sign, digits[1:] keeps the rest of the text.
````

````python
# @title Solution { display-mode: "form" }
samples = ['42', ' 42 ', '-5', '+5', '٢٠', '', '   ',
           'abc', '3.14', '-', '--5', '4 2', '1234567890123']
for raw in samples:
    clean = raw.strip()
    digits = clean
    if digits.startswith('+') or digits.startswith('-'):
        digits = digits[1:]
    if 1 <= len(digits) <= 12 and digits.isdecimal():
        print(int(clean))         # convert the same cleaned text we checked
    else:
        print('Invalid')
````

#### Slide 9A · Assignment operators (the full table)

`=` is plain assignment. Every *compound* assignment follows one rule:
`x op= y` is shorthand for `x = x op y`, and the **right-hand side is evaluated first**.

**Arithmetic compound assignments** — standard maths on a variable, storing the result back.

| Operator | Name | Example | Equivalent to | Result when `x = 5` |
| --- | --- | --- | --- | --- |
| `+=` | Addition Assignment | `x += 3` | `x = x + 3` | `8` |
| `-=` | Subtraction Assignment | `x -= 3` | `x = x - 3` | `2` |
| `*=` | Multiplication Assignment | `x *= 3` | `x = x * 3` | `15` |
| `/=` | Division Assignment | `x /= 3` | `x = x / 3` — always returns a **float** | `1.6666666666666667` |
| `//=` | Floor Division Assignment | `x //= 3` | `x = x // 3` — rounds **down** to an integer | `1` |
| `%=` | Modulus Assignment | `x %= 3` | `x = x % 3` — the **remainder** | `2` |
| `**=` | Exponentiation Assignment | `x **= 3` | `x = x ** 3` — raises to a power | `125` |

*Note.* `+=` and `*=` also work on sequences such as strings and lists, where they
mean concatenation and repetition: `text += " world"`, `items *= 2`.

**Bitwise compound assignments** — these act on the **binary digits** of an integer.

| Operator | Name | Example | Equivalent to | Result when `x = 5` |
| --- | --- | --- | --- | --- |
| `&=` | Bitwise AND Assignment | `x &= 3` | `x = x & 3` | `1` |
| `\|=` | Bitwise OR Assignment | `x \|= 3` | `x = x \| 3` | `7` |
| `^=` | Bitwise XOR Assignment | `x ^= 3` | `x = x ^ 3` | `6` |
| `>>=` | Right Shift Assignment | `x >>= 1` | `x = x >> 1` | `2` |
| `<<=` | Left Shift Assignment | `x <<= 1` | `x = x << 1` | `10` |

*Note.* In Python 3.9+, `\|=` also merges one dictionary into another in place.
Recognise these five in the table; this session never needs them.

**Three of them carry this session.** `i += 1` advances a counter (slides 6-9),
`total += value` accumulates the running sum you have just seen on slide 9, and
`n -= 1` counts down towards the condition that ends a loop (slide 13).

**`/=` changes the type.** True division always produces a float, so `n /= 2`
turns `10` into `5.0`. Use `//=` when the value must stay an integer.

**The variable must already exist.** `x += 1` reads `x` before it writes it, so
a name that was never assigned raises `NameError`. That is why every counter and
every total is initialised *before* the loop.

Table source: [W3Schools · Python assignment operators](https://www.w3schools.com/python/gloss_python_assignment_operators.asp)

````python
# The seven arithmetic rows, each one starting again from x = 5
x = 5
x += 3
print("x += 3   ->", x)          # addition:       5 + 3

x = 5
x -= 3
print("x -= 3   ->", x)          # subtraction:    5 - 3

x = 5
x *= 3
print("x *= 3   ->", x)          # multiplication: 5 * 3

x = 5
x /= 3
print("x /= 3   ->", x, type(x).__name__)   # division: ALWAYS a float

x = 5
x //= 3
print("x //= 3  ->", x)          # floor division: rounds down

x = 5
x %= 3
print("x %= 3   ->", x)          # modulus:        the remainder

x = 5
x **= 3
print("x **= 3  ->", x)          # exponentiation: 5 * 5 * 5

# The right-hand side is evaluated FIRST (slide 8):
x = 5
x *= 2 + 1                       # x = 5 * (2 + 1) = 15, NOT 5 * 2 + 1 = 11
print("x *= 2 + 1 ->", x)
````

````python
# += and *= also work on sequences: concatenation and repetition
text = "hello"
text += " world"                 # same as: text = text + " world"
print(text)

line = "-"
line *= 20                       # same as: line = line * 20
print(line)

items = [1, 2]
items += [3]                     # append onto the end of the list
items *= 2                       # repeat the whole list
print(items)

# The types must match: "5" + 3 is a TypeError, and so is "5" += 3.
# Uncomment to see: TypeError: can only concatenate str (not "int") to str
# text += 3
````

````python
# The three shorthands the loops in this session are built from
i = 0
while i <= 3:                # counter up (slides 6-9)
    print("up:", i, end="  ")
    i += 1                   # i = i + 1
print()

n = 3
while n > 0:                 # counting down is what ends the loop (slide 13)
    print("down:", n, end="  ")
    n -= 1                   # n = n - 1
print()

total = 0
k = 1
while k <= 5:                # accumulator: the slide 9 sum
    total += k               # total = total + k  (slide 9)
    k += 1
print("total =", total)

# A shorthand reads the variable before it writes it, so it must exist first.
# Uncomment the next line to see: NameError: name 'missing' is not defined
# missing += 1
````

````python
# The five bitwise rows act on the BINARY digits of an integer (recognition only)
print("5 is", format(5, "04b"), "and 3 is", format(3, "04b"), "in binary")

x = 5
x &= 3          # 0101 & 0011 -> 0001 : a 1 only where BOTH bits are 1
print("x &= 3   ->", x)

x = 5
x |= 3          # 0101 | 0011 -> 0111 : a 1 where EITHER bit is 1
print("x |= 3   ->", x)

x = 5
x ^= 3          # 0101 ^ 0011 -> 0110 : a 1 only where the bits DIFFER
print("x ^= 3   ->", x)

x = 5
x >>= 1         # 0101 shifted 1 place right -> 0010 : halves the number
print("x >>= 1  ->", x)

x = 5
x <<= 1         # 0101 shifted 1 place left  -> 1010 : doubles the number
print("x <<= 1  ->", x)

# Python 3.9+: |= also merges a dictionary into another one, in place
marks = {"ali": 8, "sara": 9}
marks |= {"omar": 7, "ali": 10}      # new key added, existing key overwritten
print(marks)
````

#### Slide 21A · break vs continue vs pass

All three are written inside a loop body, but only two of them steer the loop.

| Statement | What it does | Where the next line runs | Rest of this pass | Loop keeps going? |
| --- | --- | --- | --- | --- |
| `break` | Ends the loop immediately | The first line **after** the loop | Skipped | **No** — and the loop's `else` is skipped too |
| `continue` | Ends this pass only | Back at the loop's **condition** | Skipped | **Yes**, with the next value |
| `pass` | Nothing at all | The **very next line**, as if it were not written | Runs normally | **Yes**, untouched |

**Indentation decides which loop.** `break` and `continue` act on the *innermost*
loop that encloses them. In the demos below, the same statement two levels deep
stops only the inner loop, while the outer loop carries on.

**`pass` is not a loop statement.** It is a placeholder that keeps an empty block
legal (slide 21), so it is the only one of the three that is also allowed inside
an `if` or `else` that has nothing to do yet.

**While-loop trap.** `continue` jumps straight back to the condition, so a counter
update written *below* it never runs and the loop freezes (slide 13). The demos
update the counter on the first line of the body, where `continue` cannot skip it.

<span class="cosc-practice-slide"></span>


````python
# The SAME nested while loop three times — only the marked statement changes
print("with break:")
row = 0
while row < 3:                     # outer loop
    row += 1
    print(f"  row {row}:", end=" ")
    col = 0
    while col < 3:                 # inner loop
        col += 1                   # update FIRST, so continue cannot freeze us
        if col == 2:               # the decision
            break                  # leaves the INNER loop only
        print(col, end=" ")
    print("-> outer loop is still running")

print("\nwith continue:")
row = 0
while row < 3:
    row += 1
    print(f"  row {row}:", end=" ")
    col = 0
    while col < 3:
        col += 1
        if col == 2:
            continue               # skips the print for col 2, stays in the loop
        print(col, end=" ")
    print("-> outer loop is still running")

print("\nwith pass:")
row = 0
while row < 3:
    row += 1
    print(f"  row {row}:", end=" ")
    col = 0
    while col < 3:
        col += 1
        if col == 2:
            pass                   # does nothing: col 2 is printed anyway
        print(col, end=" ")
    print("-> outer loop is still running")
````

````python
# Five levels of indentation: count the columns to see what each statement ends
names = ["ali", "sara", "omar", "stop", "zaid"]
i = 0
while i < len(names):                          # level 1: outer loop
    name = names[i]
    i += 1
    if name == "stop":                         # level 2: inside the outer loop
        print("sentinel found -> this break ends the OUTER loop")
        break                                  # level 3: ends the whole scan
    if name.startswith("s"):                   # level 2
        tries = 0
        while tries < 3:                       # level 3: inner loop
            tries += 1
            if tries == 2:                     # level 4
                break                          # level 5: ends the INNER loop only
            print(f"  {name}: inner try {tries}")
        print(f"  {name}: inner loop stopped with tries = {tries}")
    else:
        pass                                   # level 3: nothing to do for this name
    print(f"finished {name}")                  # level 2: the outer loop resumes here
print("after the outer loop")
# 'zaid' is never reached: break on 'stop' ended the outer loop.
# 'finished stop' never prints either: break skips the rest of that pass.
````

#### Slide 21B · Nested loops

A loop body may hold any statement, so it may hold another loop. The **inner**
loop runs all the way through on **every single pass** of the **outer** loop:
3 outer passes × 4 inner passes = 12 runs of the inner body.

<div style="background:#ffffff;border:1px solid #d9e2ef;border-radius:10px;padding:8px;margin:6px 0;text-align:center;">

<p style="margin:6px 0 0 !important;font-size:14px !important;color:#526071 !important;">The outer loop brackets the whole inner loop. Written in C++, where braces mark the bodies that Python marks by indentation.</p>
</div>

The figure is C++ — braces instead of indentation, and `count++` doing the
counting inside the header — but the shape is what carries over. The same two
loops in Python are:

```python
count = 0
while count < 50:              # outer
    count += 1
    for i in range(25):        # inner: a full 25 passes every outer pass
        ...
```

**When you need one.** Any task of the form *for each X, do something with every
Y inside it*: for each line of a file, count how often `"the"` appears in it; for
each student, total their marks; for each row of a grid, print every column.

<div style="background:#ffffff;border:1px solid #d9e2ef;border-radius:10px;padding:8px;margin:6px 0;text-align:center;">

<p style="margin:6px 0 0 !important;font-size:14px !important;color:#526071 !important;">Nested wheels: the small one completes whole turns of its own for each single turn of the big one.</p>
</div>

<p style="font-size:14px;color:#526071;">Figures from the Oregon State University
CS 161 notes, chapter 5, <a href="https://eecs.engineering.oregonstate.edu/ecampus-video/CS161/template/chapter_5/nested.html">Nested loops, Which loop to
use?</a> — shown here for reference; the code beside them is ours, in Python.</p>

<span class="cosc-practice-slide"></span>


````python
# The figure above in Python: indentation replaces the braces
inner_runs = 0
row = 1
while row <= 3:                          # OUTER loop: 3 passes
    print(f"row {row}:", end=" ")
    col = 1
    while col <= 4:                      # INNER loop: 4 full passes EVERY time
        print(f"({row},{col})", end=" ")
        inner_runs += 1
        col += 1                         # inner counter: resets to 1 above
    print()                              # one line per pass of the OUTER loop
    row += 1                             # outer counter: runs once per row
print("inner body ran", inner_runs, "times = 3 x 4")
````

#### Slide 21C · break, and the flag that replaces it

<div style="background:#ffffff;border:1px solid #d9e2ef;border-radius:10px;padding:8px;margin:6px 0;text-align:center;">

<p style="margin:6px 0 0 !important;font-size:14px !important;color:#526071 !important;">A loop set to run 50 times that the user can leave early by typing q.</p>
</div>

`break` leaves the loop the moment it runs (slide 16). These notes add the
question the deck leaves open — **whether you should**. Their advice: a loop is
easier to read and debug when its header alone says when it stops, so let the
condition end it.

<div style="background:#ffffff;border:1px solid #d9e2ef;border-radius:10px;padding:8px;margin:6px 0;text-align:center;">

<p style="margin:6px 0 0 !important;font-size:14px !important;color:#526071 !important;">The same program without break: a flag in the while condition stops the loop.</p>
</div>

| C++ | Python |
| --- | --- |
| `while (count++ < 50)` | `while count < 50:` with `count += 1` as the first body line |
| `if (inputString == "q") break;` | `if reply == "q": break` |
| `bool continueInput = true;` | `keep_going = True` |
| `while (count++ < 50 && continueInput)` | `while count < 50 and keep_going:` |

**`break` inside a nested loop.** It ends only the loop it is written in. Put it
in the inner loop and the outer loop carries on, exactly as slide 21A shows; put
it in the outer loop and all the looping stops. To leave both from the inside,
set a flag that the outer condition tests.

<span class="cosc-practice-slide"></span>


````python
# Both figures in Python, driven by a list so nothing has to be typed
replies = ["Ali", "Sara", "q", "Omar"]   # what a student would enter

print("with break:")
count = 0
i = 0
while count < 50:
    count += 1
    reply = replies[i]
    i += 1
    if reply == "q":
        break                            # leaves the loop at once
    print("  name:", reply)

print("\nwith a flag instead of break:")
count = 0
i = 0
keep_going = True
while count < 50 and keep_going:         # the header alone says when it stops
    count += 1
    reply = replies[i]
    i += 1
    if reply == "q":
        keep_going = False               # the condition ends the loop next time
    else:
        print("  name:", reply)

# Same output, two different exits: one jumps out, one lets the condition decide.
````

#### Slide 22A · Leaving two loops at once

`break` ends **one** loop — the one it is written in (slide 21A). Python has no
`break 2` and no loop labels, so leaving the outer loop from inside the inner
one always takes one of these four.

| Way out | The shape of it | When to reach for it |
| --- | --- | --- |
| **Flag in the outer condition** | `keep_going = True` … `while ... and keep_going:` | **now** — the only one that needs nothing you have not met |
| **`else` + `continue` + `break`** | the inner loop's `else:` runs `continue`, then `break` below it | when you know loop-`else` (slide 22); common in real code, and hard to read |
| **Move the loops into a function** | `return` leaves **every** loop at once | after functions — what experienced Python programmers reach for |
| **Raise an exception** | `raise` inside, `except` outside the loops | after exceptions — it works, but it is heavy machinery for a loop |

**The flag does not stop the outer loop where it is set.** It is only read the
next time the outer **condition** is tested, so the rest of that outer pass still
runs. If nothing after the inner loop should run, `break` out of the outer loop
right after the inner one, or test the flag there.

Once you have **functions**, the third row is the one most Python programmers
reach for, because `return` leaves every loop at once with nothing to check
afterwards:

```python
# a preview only — functions come later in the course
def first_negative(grid):
    for row in range(len(grid)):
        for col in range(len(grid[row])):
            if grid[row][col] < 0:
                return (row, col)     # leaves BOTH loops, and the function
    return None
```

**Why no `break 2`?** Other languages do have it — PHP counts levels
(`break 2`), while Java, Go, Rust and Zig label the loop and break to the label.
In Python the idea is raised again and again and has never been accepted; the
[September 2025 discussion](https://discuss.python.org/t/break-from-multiple-loops-at-once/103766) this slide draws on runs the usual
course. `break 2` makes the reader count loops to find where the jump lands,
labels make them look the name up, and no example anyone brought reads better
that way than with a flag, a loop `else`, or a function. Two well-named loops in
a function beat one clever jump.

<span class="cosc-practice-slide"></span>


````python
# One job, two exits: find the first negative number in a grid of rows
grid = [[3, 5, 2],
        [7, -4, 8],
        [1, 9, 6]]

# 1) A flag in the outer condition — only tools from this session
found = None
row = 0
while row < len(grid) and found is None:     # the flag is read HERE
    col = 0
    while col < len(grid[row]):
        if grid[row][col] < 0:
            found = (row, col)
            break                            # ends the INNER loop only
        col += 1
    print(f"  outer pass {row} still finished")   # runs even on the last pass
    row += 1
print("flag     ->", found)

# 2) else + continue + break — the loop else runs only when no break happened
found = None
for row in range(len(grid)):
    for col in range(len(grid[row])):
        if grid[row][col] < 0:
            found = (row, col)
            break                            # a break SKIPS the else below
    else:
        continue                             # no break: try the next row
    break                                    # a break happened: leave the outer loop
print("else     ->", found)

# Same answer, two different exits. The third way — a function whose return
# leaves both loops — is on the slide above, waiting for the functions topic.
````

<ol><li><strong>Normal cases:</strong> positive, negative, and zero.</li><li><strong>Sentinel exit:</strong> single space <code style="background:#eef3fa;color:#173d72;border-radius:4px;padding:0.05em 0.25em;">' '</code> exits cleanly.</li><li><strong>Crash hazard 1:</strong> pressing Enter directly sends empty string <code style="background:#eef3fa;color:#173d72;border-radius:4px;padding:0.05em 0.25em;">''</code>. Notice <code style="background:#eef3fa;color:#173d72;border-radius:4px;padding:0.05em 0.25em;">''.isspace()</code> is <code style="background:#eef3fa;color:#173d72;border-radius:4px;padding:0.05em 0.25em;">False</code>, so <code style="background:#eef3fa;color:#173d72;border-radius:4px;padding:0.05em 0.25em;">int('')</code> stops the program!</li><li><strong>Crash hazard 2:</strong> non-numeric input <code style="background:#eef3fa;color:#173d72;border-radius:4px;padding:0.05em 0.25em;">'abc'</code> also stops <code style="background:#eef3fa;color:#173d72;border-radius:4px;padding:0.05em 0.25em;">int()</code>. Check the text with <code style="background:#eef3fa;color:#173d72;border-radius:4px;padding:0.05em 0.25em;">.isdecimal()</code> (Practice 3C) <em>before</em> converting it.</li><li><strong>Refactor:</strong> replace 3 independent <code style="background:#eef3fa;color:#173d72;border-radius:4px;padding:0.05em 0.25em;">if</code>s with <code style="background:#eef3fa;color:#173d72;border-radius:4px;padding:0.05em 0.25em;">if/elif/else</code>.</li></ol>
<p><strong>Target:</strong> a loop that exits on space OR empty string, ignores non-numbers safely, and classifies correctly.</p>
</div>

````python
# @title Solution { display-mode: "form" }

# Corrected robust version — one loop over canned entries instead of input()
# Normal values, invalid text, then the empty string a bare Enter produces:
entries = ["10", "abc", "-4", "0", "", "3"]
positives = negatives = zeros = 0

print("Testing invalid text, then the empty-string exit:")
for b in entries:
    # Exit on space OR empty string (user pressing Enter)
    if not b or b.isspace():
        print("Exit sentinel reached.")
        break
    # Check the text BEFORE converting it (the Practice 3C rule)
    digits = b
    if digits.startswith('+') or digits.startswith('-'):
        digits = digits[1:]
    if not digits.isdecimal():
        print(f"Warning: {b!r} is not a valid integer. Skipped.")
        continue
    c = int(b)                # safe now: every character was checked
    # Mutually exclusive branches: if/elif/else
    if c > 0:
        positives += 1
    elif c < 0:
        negatives += 1
    else:
        zeros += 1

print(f"Results: positives={positives}, negatives={negatives}, zeros={zeros}")
# The trailing "3" is never counted: the empty string ended the loop first.
````

````python
# Reverse a number (slide code)
# The idea: peel the LAST digit off n, then stick it onto the END of rev.
# Two operators do all the work:
#     n % 10   -> the last digit of n           (371 % 10  = 1)
#     n // 10  -> n WITHOUT its last digit      (371 // 10 = 37)

n = int(input("Enter a number: "))   # input() hands back text, int() turns it into a number
rev = 0                              # the answer, built one digit at a time (nothing yet)

while n > 0:                         # keep looping while n still has digits left
    r = n % 10                       # r = the digit we just peeled off the right end of n
    rev = rev * 10 + r               # *10 shifts rev one place left, + r fills the empty place
    n = n // 10                      # drop the digit we used: this is what ends the loop

print("Reversed number:", rev)       # n has reached 0, so every digit is now inside rev

# Trace for 371 (the same numbers as slide 27):
#   start           n = 371    rev = 0
#   pass 1   r = 1  n =  37    rev =  0 * 10 + 1 =   1
#   pass 2   r = 7  n =   3    rev =  1 * 10 + 7 =  17
#   pass 3   r = 3  n =   0    rev = 17 * 10 + 3 = 173   -> n is 0, so the loop stops
#
# Two questions students always ask:
#   450 reverses to 54, not 054 - an integer cannot keep a leading zero.
#   A negative number never enters the loop (n > 0 is False), so rev stays 0.
````

<h2 style="color:#2457a6;font-size:22px;margin:0 0 0.6em;">Exercise 2: Robust Min & Max program</h2>
<ol><li><strong>Fix initialization:</strong> setting <code style="background:#eef3fa;color:#173d72;border-radius:4px;padding:0.05em 0.25em;">largest = 0</code> fails on all-negative inputs. Use <code style="background:#eef3fa;color:#173d72;border-radius:4px;padding:0.05em 0.25em;">None</code> or read the first number to seed both.</li><li><strong>Dual tracking:</strong> update both <code style="background:#eef3fa;color:#173d72;border-radius:4px;padding:0.05em 0.25em;">smallest</code> and <code style="background:#eef3fa;color:#173d72;border-radius:4px;padding:0.05em 0.25em;">largest</code> on each valid input.</li><li><strong>Input validation:</strong> check each entry with <code style="background:#eef3fa;color:#173d72;border-radius:4px;padding:0.05em 0.25em;">.isdecimal()</code> before converting, so non-numeric text is skipped instead of stopping the loop.</li></ol>
<p><strong>Target:</strong> reports both smallest and largest correctly, even when all numbers are negative.</p>
</div>
</details>

````python
# @title Solution { display-mode: "form" }

# All-negative numbers, one invalid entry, the 'a' sentinel, and a value after it:
inputs = ["-15", "-3", "bad", "-28", "-7", "a", "99"]
smallest = None
largest = None

for entry in inputs:
    if entry == 'a':
        print("Sentinel 'a' reached — stopping.")
        break

    # Check the text BEFORE converting it (the Practice 3C rule)
    digits = entry
    if digits.startswith('+') or digits.startswith('-'):
        digits = digits[1:]
    if not digits.isdecimal():
        print(f"Skipping non-integer input: {entry!r}")
        continue
    num = int(entry)

    if smallest is None or num < smallest:
        smallest = num
    if largest is None or num > largest:
        largest = num

if smallest is not None:
    print(f"Smallest = {smallest}, Largest = {largest}")
else:
    print("No valid numbers were entered.")
# largest = -3, not 0: starting from None is what makes all-negative input work.
# "99" after the sentinel is never read — break left the loop immediately.
````
