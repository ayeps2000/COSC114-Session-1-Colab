#### Slide 15 &middot; live code below

<div style="background:#ffffff;border:1px solid #d9e2ef;border-radius:10px;padding:6px;margin:2px 0 4px;text-align:center;"><img alt="PowerPoint slide 15: Slicing with a stride" loading="lazy" decoding="async" src="https://raw.githubusercontent.com/ayeps2000/COSC114-Session-1-Colab/main/slides/ss6/slide-15.png" style="display:block;margin:0 auto;width:auto;max-width:100%;max-height:62vh;"></div>

````python
mylist = ["cup", 20, 2.3, "football", True, "cup"]
print(mylist[1:6])     # default stride = 1
print(mylist[1:6:1])   # stride = 1
print(mylist[1:6:2])   # stride = 2, takes every second item
````

<div style="background:#f2f8f5;border-left:6px solid #198754;border-radius:8px;padding:10px 16px;"><p style="color:#198754;font-weight:700;margin:0 0 0.4em;">Answer</p><pre style="background:white;border:1px solid #d9e2ef;border-radius:6px;padding:8px 12px;font-family:Menlo,Consolas,monospace;font-size:18px;white-space:pre-wrap;">[20, 2.3, &#x27;football&#x27;, True, &#x27;cup&#x27;]
[20, 2.3, &#x27;football&#x27;, True, &#x27;cup&#x27;]
[20, &#x27;football&#x27;, &#x27;cup&#x27;]</pre><p style="margin:0;color:#526071;">The third number is the stride: how far to step after each item. Stride 2 takes indexes 1, 3 and 5.</p></div>


#### Slide 16 &middot; live code below

<div style="background:#ffffff;border:1px solid #d9e2ef;border-radius:10px;padding:6px;margin:2px 0 4px;text-align:center;"><img alt="PowerPoint slide 16: More strides" loading="lazy" decoding="async" src="https://raw.githubusercontent.com/ayeps2000/COSC114-Session-1-Colab/main/slides/ss6/slide-16.png" style="display:block;margin:0 auto;width:auto;max-width:100%;max-height:62vh;"></div>

````python
mylist = ["cup", 20, 2.3, "football", True, "cup"]
print(mylist[0:10:2])   # an end past the last index just stops at the end
print(mylist[1:6:5])
print(mylist[::2])
print(mylist[-6:-1:3])
````

[&#x27;cup&#x27;, &#x27;football&#x27;]</pre><p style="margin:0;color:#526071;">Slicing never raises IndexError: an end of 10 just stops at the last item. [1:6:5] takes index 1, and the next step (6) is already past the end.</p></div>


````python
mylist = ["cup", 20, 2.3, "football", True, "cup"]
print(mylist[::-1])
````

````python
# @title Solution { display-mode: "form" }
grades = {"Ali": 85, "Sara": 90, "Omar": 89}

for name in ["Sara", "Mona"]:
    if name in grades:
        print(name + ":", grades[name])
    else:
        print(name + ": not found")

for name in grades:
    print(name, grades[name])
````
