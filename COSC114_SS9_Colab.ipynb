## Strings vs. Lists & Tuples: Don't Repeat What You Already Know

> **The 30-Second Rule:** *"A string is an **immutable sequence of characters**. Everything that reads a list or tuple (indexing `s[0]`, negative indices `s[-1]`, slicing `s[1:4]`, `len()`, `in`, and iterating with loops) works on strings. What makes strings unique is that they **cannot be modified in place**, they have specialized **text methods** (`split`, `strip`, `replace`, `join`), and they use modern **f-strings** for high-precision output formatting."*

### 1. Works the same as Lists & Tuples (Point back to SS7 & SS8 — Skim)

| Feature | Strings (SS9) | Lists (SS7) & Tuples (SS8) | Slide |
|---|---|---|---|
| Indexing & negative indices | `s[0]`, `s[-1]` | Identical zero-based indexing | Slide 2 |
| Slicing | `s[1:4]`, `s[::-1]` | Identical `[start:stop:step]` mechanics | Slide 2, 9 |
| Length | `len(s)` | Returns total number of characters | Slide 2 |
| Concatenation & Repetition | `s1 + s2`, `s * 3` | Joins or repeats elements into a new object | Slides 3, 8 |
| Membership & Loops | `'py' in s`, `for ch in s:` | Sequences iterate element by element | Slides 5, 7 |
| Sequence search | `s.count()`, `s.index()` | Present on all sequence types | Slides 24, 27 |

### 2. Actually different (Focus class time here)

| Topic | Strings (SS9) | Lists (SS7) | Tuples (SS8) | Slides |
|---|---|---|---|---|
| **Element Type** | **Characters only** (length-1 str) | Any arbitrary objects | Any arbitrary objects | Slide 2 |

| **Trimming** | `.strip()`, `.lstrip()`, `.rstrip()` | N/A | N/A | Slides 34–36 |
| **Joining** | `sep.join(list_of_strings)` | N/A | N/A | Slide 6 |

## Actually Different: Strings are Strictly Immutable

Unlike a list, individual characters inside a string **cannot be changed**.
Any method or operation that appears to modify a string actually creates a **brand new string**.


#### Slide 11

<div style="background:#ffffff;border:1px solid #d9e2ef;border-radius:10px;padding:6px;margin:2px 0 4px;text-align:center;"><img alt="SS9 slide 11: Actually Different: String += Creates a New Object (id Changes)" loading="lazy" decoding="async" src="https://raw.githubusercontent.com/ayeps2000/COSC114-Session-1-Colab/main/slides/ss9/slide-11.png" style="display:block;margin:0 auto;width:auto;max-width:100%;max-height:62vh;"></div>

When you write `s += ' Bye'`, Python creates a **new** string `'Hello Bye'` in memory and assigns it to `s`. The original string `'Hello'` is left untouched (and garbage-collected if no other variable references it).

### f-String Format Specifier Mini-Language: `f"{value:width.precision type}"`

| Component | Example | Meaning |
|---|---|---|
| **Width** | `{text:13s}` | Minimum total characters displayed (padded with spaces) |
| **Alignment** | `{text:<13}`, `{num:>6}`, `{s:^10}` | `<` left (default for str), `>` right (default for numbers), `^` center |
| **Float Precision** | `{val:.2f}` | Number of digits after the decimal point (rounds!) |
| **String Precision** | `{text:.4s}` | Maximum number of characters displayed (truncates!) |
| **Zero-Padding** | `{num:06d}` | Pads numbers with leading zeros instead of spaces |

```python
def interleave(s1, s2):
    res = []
    min_len = min(len(s1), len(s2))
    for i in range(min_len):
        res.append(s1[i])
        res.append(s2[i])
    res.append(s1[min_len:])
    res.append(s2[min_len:])
    return "".join(res)

print(interleave("abc", "12345"))
print(interleave("Python", "KU"))
```

```python
# @title Solution { display-mode: "form" }
def check_password(pw):
    has_letter = any(c.isalpha() for c in pw)
    has_digit = any(c.isdigit() for c in pw)
    has_special = any(not c.isalnum() for c in pw)
    if len(pw) >= 8 and has_letter and has_digit and has_special:
        return "Password is strong"
    elif len(pw) >= 8 and has_letter and has_digit:
        return "Password is medium"
    else:
        return "Password is weak"

print(check_password("abc"))
print(check_password("password123"))
print(check_password("Pass123!"))
```

### Session 9 Summary & Key Takeaways

1. **Strings are Immutable Sequences:** No item assignment (`s[0] = ...`), no `.append()`. `+=` creates a new string.
2. **f-Strings (`f"{val:...}"`):** Width pads, `<` left-aligns, `>` right-aligns, `.2f` rounds floats, `.4s` truncates strings. Remember: precision on integers (`:.3d`) is **forbidden**!
3. **`.split()` vs `.split(sep)`:** Default collapses whitespace; explicit preserves empty strings `''`.
4. **`.strip(chars)`:** Removes from both ends until a non-matching character is found. Outer characters block inner ones.
5. **`.find()` vs `.index()`:** Use `.find()` when a missing substring should return `-1` rather than crashing.
