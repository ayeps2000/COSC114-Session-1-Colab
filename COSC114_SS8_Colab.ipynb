```python
# @title Solution { display-mode: "form" }
def passing_names(names, scores, min_score):
    passed = []
    for name, score in zip(names, scores):
        if score >= min_score:
            passed.append(name)
    return tuple(passed)

names = ("Alice", "Bob", "Charlie", "David")
scores = (85, 62, 90, 70)
print(passing_names(names, scores, 70))
assert passing_names(names, scores, 70) == ("Alice", "Charlie", "David")
```
